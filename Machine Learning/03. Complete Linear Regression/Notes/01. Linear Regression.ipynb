# 📘 In-Depth Tutorial on Linear Regression

---

## 🟧 Definition: What is Linear Regression?

> **Linear Regression** is a supervised learning algorithm used for predicting a quantitative (continuous) response using one or more input features by fitting a linear relationship between input and output.



## 🟧 Types of Linear Regression

| Type                           | Description                                                          |
| ------------------------------ | -------------------------------------------------------------------- |
| **Simple Linear Regression**   | One input (independent) variable and one output (dependent) variable |
| **Multiple Linear Regression** | Multiple input variables predicting a single output                  |



## 🟧 Equation of Linear Regression

### 🔹 Simple Linear Regression

$y = \beta_0 + \beta_1x + \epsilon$

* $y$: predicted output  
* $x$: input feature  
* $\beta_0$: intercept  
* $\beta_1$: slope (coefficient)  
* $\epsilon$: error term  

### 🔹 Multiple Linear Regression

$y = \beta_0 + \beta_1x_1 + \beta_2x_2 + ... + \beta_nx_n + \epsilon$



## 🟧 Cost Function (Loss Function)

> The cost function measures how well the model's predictions match actual data.

### 🔹 Mean Squared Error (MSE)

$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m}(\hat{y}_i - y_i)^2$

* $m$: number of samples  
* $\hat{y}_i$: predicted output  
* $y_i$: actual output  



## 🟧 Optimization: Gradient Descent

> Gradient Descent is an iterative method to minimize the cost function and find the best values of $\beta$.

### 🔹 Gradient Descent Update Rule

$\beta_j = \beta_j - \alpha \frac{\partial J}{\partial \beta_j}$

* $\alpha$: learning rate  



## 🟧 Assumptions of Linear Regression

1. **Linearity**: The relationship between the input and output is linear.  
2. **Independence**: Observations are independent.  
3. **Homoscedasticity**: Constant variance of error terms.  
4. **Normality of Errors**: Error terms are normally distributed.  
5. **No Multicollinearity**: Independent variables are not highly correlated with each other.  



## 🟧 Model Evaluation Metrics

| Metric                | Formula                             | Use                        |
| --------------------- | ----------------------------------- | -------------------------- |
| **R-squared** ($R^2$) | $1 - \frac{SS_{res}}{SS_{tot}}$     | Measures goodness of fit   |
| **MSE / RMSE**        | Average of squared (or root) errors | Penalizes large errors     |
| **MAE**               | Mean of absolute errors             | Less sensitive to outliers |



## 🟧 Implementation in Python (Scikit-learn)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Training Data
X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)

y_pred = model.predict(X)
print("R^2:", r2_score(y, y_pred))
```

## 🟧 Pros and Cons

### 🔹 Advantages:

* Simple to understand and implement  
* Fast training  
* Works well for linearly separable data  

### 🔹 Limitations:

* Assumes linearity  
* Sensitive to outliers  
* Prone to underfitting on complex data  



## 🟧 Interview Questions & Answers

#### **❓ What is linear regression?**

 It’s a technique to model the relationship between dependent and independent variables by fitting a linear equation.

#### **❓ What is the cost function in linear regression?**

 Mean Squared Error (MSE), which measures the average squared difference between actual and predicted values.

#### **❓ What are the assumptions of linear regression?**

 Linearity, Independence, Homoscedasticity, Normality of Errors, and No Multicollinearity.

#### **❓ What is R-squared?**

 It measures the proportion of the variance in the dependent variable that is predictable from the independent variables.

#### **❓ How is gradient descent used in linear regression?**

 It minimizes the cost function by updating weights in the direction of the negative gradient.

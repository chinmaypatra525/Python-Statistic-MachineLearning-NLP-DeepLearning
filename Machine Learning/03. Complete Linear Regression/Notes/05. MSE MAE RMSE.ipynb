# 📘 MSE, MAE, RMSE in Linear Regression

---

## 🟧 Introduction

In supervised learning, especially regression tasks, we need to quantify how well our model performs — i.e., how far off the predictions ($\hat{y}$) are from the actual values ($y$).

Three common error metrics used are:

- **MAE**: Mean Absolute Error  
- **MSE**: Mean Squared Error  
- **RMSE**: Root Mean Squared Error  

These are all types of **loss functions** or **performance metrics** that help evaluate and optimize regression models.



## 🟧 1. Mean Absolute Error (MAE)

### 🔹 Intuition:
MAE tells you the **average magnitude** of errors between predicted and actual values, **without considering direction** (i.e., all errors are positive).

### 🔹 Formula:
$$
MAE = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|
$$

### 🔹 Characteristics:
- Linear penalty: Each error contributes proportionally.  
- Does not square the error — less sensitive to outliers.  
- The best line minimizes the **median** of the errors.

### 🔹 Pros:
- Easy to understand  
- Robust to outliers  
- Same unit as target variable  

### 🔹 Cons:
- Not differentiable at zero (makes optimization harder)  
- Doesn’t emphasize large errors  



## 🟧 2. Mean Squared Error (MSE)

### 🔹 Intuition:
MSE measures the **average of squared errors**, giving larger weight to larger errors.

### 🔹 Formula:
$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

### 🔹 Characteristics:
- Squared penalty: Large errors have exponentially larger impact.  
- Minimizing MSE leads to **least squares regression**.  
- The best line minimizes the **mean of squared errors**.

### 🔹 Pros:
- Differentiable and smooth → good for optimization (e.g., gradient descent)  
- Heavily penalizes large errors → good if large errors are very undesirable  

### 🔹 Cons:
- Sensitive to outliers  
- Unit is squared (e.g., if target is in dollars, MSE is in dollars²)  



## 🟧 3. Root Mean Squared Error (RMSE)

### 🔹 Intuition:
RMSE is simply the **square root of MSE**, which brings the error back to the original scale of the target variable.

### 🔹 Formula:
$$
RMSE = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 }
$$

### 🔹 Characteristics:
- Directly interpretable in the same units as the target variable  
- Like MSE, it penalizes larger errors more

### 🔹 Pros:
- More interpretable than MSE  
- Smooth and differentiable  
- Popular in competitions (e.g., Kaggle)

### 🔹 Cons:
- Still sensitive to outliers



## 🟧 Comparison Table

| Metric | Formula | Unit | Penalizes Large Errors? | Outlier Sensitivity | Use Case |
|--------|---------|------|--------------------------|----------------------|----------|
| **MAE** | $\frac{1}{n} \sum \left| y_i - \hat{y}_i \right|$ | Original Unit | ❌ No | Low | Robust predictions |
| **MSE** | $\frac{1}{n} \sum (y_i - \hat{y}_i)^2$ | Squared Unit | ✅ Yes | High | Optimization, theory |
| **RMSE** | $\sqrt{MSE}$ | Original Unit | ✅ Yes | High | Intuitive comparison |



## 🟧 Which One Should You Use?

| Scenario | Recommended Metric |
|----------|---------------------|
| Model needs to be robust to outliers | **MAE** |
| You want to heavily penalize large errors | **MSE** or **RMSE** |
| You need an interpretable metric (same unit as target) | **RMSE** or **MAE** |
| You care about optimization and theoretical convenience | **MSE** (used in gradient descent) |



## 🟧 Bonus: MSE in Gradient Descent

The derivative of the MSE cost function w.r.t weights $w$ is:

$$
\frac{d}{dw} MSE = \frac{2}{n} \sum (y_i - \hat{y}_i)(-x_i)
$$

This is used in **gradient descent** to update the weights in linear regression.

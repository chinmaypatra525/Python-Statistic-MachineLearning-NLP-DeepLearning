# 📘 In-Depth Tutorial on Cross-Validation in Machine Learning

---

## 🟧 Definition: What is Cross-Validation?

> **Cross-validation** is a resampling technique used to assess how well a model will generalize to an independent dataset. It helps in reducing overfitting and provides insights into model performance.



## 🟧 Why Use Cross-Validation?

* Ensures that the model’s performance is consistent across different subsets of data.
* Prevents overfitting by verifying that the model does not memorize the training set.
* Useful for **model evaluation**, **model selection**, and **hyperparameter tuning**.



## 🟧 Basic Idea

> Split the original dataset into multiple subsets (folds), train the model on some of the subsets, and validate it on the remaining ones.



## 🟧 Types of Cross-Validation

### 🔹 1. **Hold-Out Validation**

* Splits the data into **training and test** sets (e.g., 80/20 or 70/30).
* Simple but can lead to high variance.

<br>
<img src="./Images/Hold-Out Validation.jpg" alt="Hold-Out Validation" width="700"/>
<br>

### 🔹 2. **K-Fold Cross-Validation**

> The dataset is divided into `k` equal-sized folds. The model is trained `k` times, each time leaving out one different fold as the validation set.

<br>
<img src="./Images/Cross Validation.webp" alt="Cross Validation" width="700"/>
<br>

* Typically, `k=5` or `k=10`.
* Reduces bias and variance.

### 🔹 3. **Stratified K-Fold Cross-Validation**

> Variation of K-Fold where each fold contains approximately the **same class distribution** as the original dataset (used for classification problems).

<br>
<img src="./Images/8-Stratified-K-Fold-Cross-Validation.png" alt="8-Stratified-K-Fold-Cross-Validation" width="700"/>
<br>



### 🔹 4. **Leave-One-Out Cross-Validation (LOOCV)**

> Each sample is used once as a test set while the rest are training samples.

* Very high computational cost.
* Best for small datasets.


<br>
<img src="./Images/leave one out cross.png" alt="leave one out cross" width="700"/>
<br>

### 🔹 5. **Repeated K-Fold**

> Applies K-Fold cross-validation **multiple times** with different random splits.

<br>
<img src="./Images/Repeated K-Fold.jpg" alt="Repeated K-Fold" width="700"/>
<br>

* Reduces variability of results.

### 🔹 6. **TimeSeriesSplit (Rolling Window)**

> Used for time-series data. Ensures the validation set is always ahead in time compared to training.



## 🟧 How to Implement K-Fold in Scikit-Learn

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_squared_error

X, y = load_diabetes(return_X_y=True)
kf = KFold(n_splits=5)
model = LinearRegression()

scores = cross_val_score(model, X, y, cv=kf, scoring='neg_mean_squared_error')
print("MSE per fold:", -scores)
print("Average MSE:", -scores.mean())
```



## 🟧 Cross-Validation vs Train/Test Split

| Aspect      | Train/Test Split | Cross-Validation |
| ----------- | ---------------- | ---------------- |
| Simplicity  | Simple           | More complex     |
| Bias        | High             | Low              |
| Variance    | High             | Low              |
| Computation | Fast             | Slower           |



## 🟧 When to Use Which Type?

| Scenario           | Recommended CV     |
| ------------------ | ------------------ |
| Large dataset      | K-Fold (k=5 or 10) |
| Imbalanced classes | Stratified K-Fold  |
| Small dataset      | LOOCV              |
| Time-series data   | TimeSeriesSplit    |



## 🟧 Common Interview Questions

#### **❓ What is cross-validation?**
**✅ Answer:**

 A technique to evaluate the generalization performance of a model by splitting data into multiple folds.

#### **❓ What is the difference between K-Fold and Stratified K-Fold?**
**✅ Answer:**

 K-Fold splits the data randomly into k subsets; Stratified K-Fold also ensures class proportions are preserved.

#### **❓ What are the disadvantages of LOOCV?**
**✅ Answer:**

 It is computationally expensive and can lead to high variance.

#### **❓ Why is cross-validation better than a simple train-test split?**
**✅ Answer:**

 It uses all the data for both training and testing, reducing bias and giving a more stable estimate of model performance.

#### **❓ What type of cross-validation should be used for time-series data?**
**✅ Answer:**

 TimeSeriesSplit, as it respects the temporal ordering of data.

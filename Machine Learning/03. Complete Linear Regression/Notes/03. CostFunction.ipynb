# 📘 Linear Regression: Cost Function, R², and Adjusted R²

---

## 🔸 What is Linear Regression?

Linear regression is a type of math model used to predict a value (like a student’s test score) based on a line (like how many hours they studied).

In simple terms:

- Linear Regression tries to draw the best straight line through a group of data points.
- That line is used to predict new values.


<br>

## 🔸 What is a Cost Function?

A **cost function** tells us how wrong our model is — it measures the error between the predicted values (from the line) and the actual values (real data).

🎯 **Goal**: Make this cost as small as possible.

> Imagine you're trying to guess someone’s marks. The cost function tells you how far your guesses are from the real marks.


<br>

## 🔸 Mean Squared Error (MSE): The Most Common Cost Function

### 🔹 What is MSE?

MSE (Mean Squared Error) finds the average of all the **squared differences** between the actual value and the predicted value.

### 🔹 Formula:

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Where:

- $y_i$: Actual value (real answer)  
- $\hat{y}_i$: Predicted value (what the model says)  
- $n$: Total number of data points  



### 🔹 Why "squared"?

- Squaring keeps all errors positive (negatives cancel out).
- Bigger mistakes get punished more.  
  (e.g., a mistake of 3 becomes 9!)


<br>

## 🔸 Sum of Squared Errors (RSS or SSE)

Before averaging like in MSE, the **total of all squared errors** is:

$$
\text{RSS} = \sum (y_i - \hat{y}_i)^2
$$

Also known as:
- Residual Sum of Squares (RSS)
- Sum of Squared Errors (SSE)

📌 Think of RSS as the total "mistake energy" of the model.


<br>

## 🔸 How Do We Predict Values?

In simple linear regression:

$$
\hat{y}_i = \theta_0 + \theta_1 x_i
$$

Where:
- $\theta_0$: y-intercept (where the line hits y-axis)  
- $\theta_1$: slope (how much y changes as x changes)  
- $x_i$: the input (like hours studied)


<br>

## 🔸 Cost Function in Linear Regression

We use **MSE** to build the cost function:

$$
J(\theta_0, \theta_1) = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - (\theta_0 + \theta_1 x_i) \right)^2
$$

This is the function we want to minimize.


<br>

## 🔸 How Do We Minimize the Cost?

We use a method called **Gradient Descent** — a smart way of slowly changing values of $\theta_0$ and $\theta_1$ to reduce the cost.



<br>



## 🔸 What is Gradient Descent?

**Gradient Descent** is an algorithm used to minimize the cost function and find the best parameters (like slope and intercept) in a machine learning model.

📉 Imagine you're at the top of a hill (high cost) and want to reach the bottom of the valley (lowest cost). Gradient descent helps you take small steps downhill to reach that lowest point.



### 🔹 Why Do We Need Gradient Descent?

In linear regression, we try to find the best line:

$ \hat{y} = \theta_0 + \theta_1 x $

We need to find the optimal values of:
- $ \theta_0 $ (intercept)  
- $ \theta_1 $ (slope)  

that **minimize the cost function** (usually MSE).  
**Gradient Descent** is the algorithm used to achieve this.



### 🔹 How Does It Work?

1. **Initialize** $ \theta_0 $ and $ \theta_1 $ with random values  
2. **Calculate the cost** (error) using the cost function (e.g., MSE)  
3. **Compute the slope** (gradient) of the cost function  
4. **Update the parameters** by taking a small step in the direction that reduces the cost  
5. **Repeat** until the cost stops changing meaningfully (*convergence*)

You keep iterating until:

- The cost function converges (changes are very small),  
- Or you reach a predetermined number of iterations.


### 🔹 Intuitive Summary

- **Think**: Starting at the top of the hill (high cost).  
- **Goal**: Walk downhill in small steps to reach the valley (global minimum).  
- **Gradient** tells you which direction is downhill.  
- **Learning rate** controls your step size—too big → you might overshoot, too small → learning is slow.


### 🔹 Gradient Descent Formula

For each parameter $ \theta_j $:

$$
\theta_j := \theta_j - \alpha \cdot \frac{\partial J}{\partial \theta_j}
$$


Where:

- $ \theta_j $: The parameter being updated (either $ \theta_0 $ or $ \theta_1 $)  
- $ \alpha $: Learning rate (controls step size)  
- $ \frac{\partial J}{\partial \theta_j} $: Derivative (slope) of the cost function with respect to $ \theta_j $



### 🔹 Visual Analogy

🧠 Imagine the **cost function** as a big bowl. You place a ball on the side of the bowl and let it roll down:

- If the **slope is steep** → the ball rolls faster (big step)  
- If the **slope is flat** → the ball rolls slowly (small step)  
- 🎯 The goal is to stop exactly at the bottom (minimum cost)

### 🔹 Learning Rate $ \alpha $

This value is **critical** in gradient descent. It determines how big your step is during each update.

### 🔹 What Happens Based on $ \alpha $:

| $ \alpha $ Value     | Behavior                                    |
|----------------------|---------------------------------------------|
| Too Small            | Very slow convergence (takes forever)       |
| Too Large            | Overshoots the minimum or diverges          |
| Just Right ✅        | Smooth, efficient convergence to minimum     |

## 🔸 What Does the Cost Surface Look Like?

- The cost function graph is **bowl-shaped (convex)**.
- It has only **one lowest point** — the global minimum.
- We slide down this bowl using gradient descent.

# 📘 Now Let’s Talk About R² (R-squared)

## 🔸 What is R²?

**R-squared** tells us how good our model is.

It compares our model to a very basic one that guesses the average every time.

### 🔹 Formula:

$$
R^2 = 1 - \frac{\text{RSS}}{\text{TSS}}
$$

Where:

- **RSS**: Residual Sum of Squares (model error)
- **TSS**: Total Sum of Squares (variation from mean)

It tells us:  
> “My model explains this much of the total variation.”


<br>

## 🔸 TSS Explained

$$
\text{TSS} = \sum (y_i - \bar{y})^2
$$

Where:

- $\bar{y}$: Mean of all actual values

It shows how much the data varies **without any model**.



### 🔹 Example:

Let:

- $TSS = 100$
- $RSS = 20$

Then:

$$
R^2 = 1 - \frac{20}{100} = 0.8
$$

📌 The model explains **80%** of the variance.


<br>

## 🔸 R² vs Cost Function

| Comparison Point | Cost Function (MSE)       | R² (R-squared)           |
|------------------|---------------------------|--------------------------|
| What it measures | Avg. of squared errors    | % of variation explained |
| Goal             | Minimize                  | Maximize                 |
| Good value       | Closer to 0               | Closer to 1              |
| Units            | Squared units             | No units                 |
| Use in training  | ✅ Yes                    | ❌ No (evaluation only)  |


<br>

## 🔸 Adjusted R² – For Multiple Features

If you add more features, R² can increase **even if features are useless**.  
That's why we use **Adjusted R²**.

### 🔹 Formula:

$$
\text{Adjusted } R^2 = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - k - 1} \right)
$$

Where:

- $n$: Number of observations  
- $k$: Number of features  



### 🔹 Why use it?

- Punishes unnecessary features.
- Increases **only if** the new feature improves the model.


<br>

## 🔸 Final Summary

| Metric        | What It Shows                      | Good Value        |
|---------------|-------------------------------------|-------------------|
| MSE           | Average squared error               | Close to 0        |
| RSS           | Total error (not averaged)          | Lower is better   |
| R²            | % of variation explained            | Close to 1        |
| Adjusted R²   | R² with penalty for extra features  | Close to 1        |


<br>

## 🔸 Key Takeaways

- Linear regression fits a straight line to data.
- We use **MSE** as the cost function to train the model.
- We use **R²** and **Adjusted R²** to evaluate the model.
- **Lower cost** = better training  
- **Higher R²** = better evaluation  
- **Adjusted R²** is better when you have **many input variables**.

  


# 📘 Polynomial Regression in Machine Learning

---

## 🟧 What is Polynomial Regression?

Polynomial Regression is an extension of Linear Regression where the relationship between the independent variable `x` and the dependent variable `y` is modeled as an `nᵗʰ`-degree polynomial.

> It is still a **linear model** in terms of coefficients, but **non-linear** in terms of input variables.



## 🟧 General Form

For degree `n`, the equation is:

$$
y = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \dots + \beta_n x^n + \varepsilon
$$


Where:

- $y$: target variable  
- $x$: feature/input variable  
- $\beta_i$: coefficients to learn  
- $\varepsilon$: error term  




## 🟧 Why Use Polynomial Regression?

| Situation                 | Explanation                                      |
|--------------------------|--------------------------------------------------|
| Data shows non-linear trends | Linear models underfit curved relationships |
| You want a flexible model    | Higher-degree polynomials can fit complex curves |
| Used for curve fitting tasks | Especially in physics, economics, etc.       |



## 🟧 How It Works (Step-by-Step)

1. **Start with Linear Regression**  
   $$
   y = \beta_0 + \beta_1 x
   $$

2. **Add Polynomial Features**  
   $$
   y = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \dots
   $$

3. **Treat each power of `x` as a separate feature**  
   Input `x` becomes:  
   $$
   [x, x^2, x^3, \dots, x^n]
   $$


4. **Apply Linear Regression** on the transformed features.



## 🟧 Choosing the Degree

| Degree         | Fit           | Risk        |
|----------------|---------------|-------------|
| Low (e.g., 1–2) | Underfitting  | Too simple  |
| High (e.g., 6+) | Overfitting   | Too complex |
| Just Right     | Good generalization | Balance |

> 🔹 Use **Cross Validation** or **Regularization (Ridge/Lasso)** to choose the optimal degree.



## 🟧 Difference from Linear Regression

| Feature           | Linear Regression                  | Polynomial Regression                                |
|-------------------|------------------------------------|------------------------------------------------------|
| Relationship      | Straight line                      | Curve                                                |
| Equation          | $y = \beta_0 + \beta_1 x$          | $y = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots$      |
| Input Features    | Original features                  | Polynomial-transformed features                      |
| Underfitting Risk | High on curves                     | Less if degree chosen well                           |




## 🟧 Summary Table

| Term                | Meaning                                                         |
|---------------------|-----------------------------------------------------------------|
| Degree              | The highest power of \( x \) in the polynomial                  |
| Overfitting         | Too high a degree fits noise, not just signal                  |
| Underfitting        | Too low a degree fails to capture pattern                      |
| `PolynomialFeatures` | sklearn class to expand input to polynomial terms             |



## 🟧 Cautions

- Too high a degree → **overfitting**
- **Numerical instability**: large powers of `x` can lead to huge values
- Not suitable for **extrapolation**: polynomials explode quickly outside the data range



## 🟧 Bonus: Polynomial Regression with Multiple Features

If you have more than one feature, `PolynomialFeatures` will generate combinations like:

$[x_1, x_2, x_1^2, x_2^2, x_1 x_2, \dots]$

## 🟧 What is Regularization?

**Definition:** Regularization is a technique used to reduce model overfitting by penalizing large coefficients in regression models.

It helps keep the model generalized, especially when:

    - Dataset is small or noisy
    - There are too many features
    - Multicollinearity exists (highly correlated features)


## 🟧 Ridge Regression (L2 Regularization)

**Definition:** Ridge Regression is a regularized version of linear regression that adds an L2 penalty (squared magnitude of coefficients) to the loss function to shrink coefficient values.

### 🔹 Ridge Intuition:

Instead of minimizing only the Sum of Squared Errors (SSE) like in Linear Regression, Ridge minimizes:

**Cost Function:**

$$
J(\theta) = \underbrace{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}_{\text{SSE}} + \underbrace{\lambda \sum_{j=1}^{p} \theta_j^2}_{\text{Penalty Term}}
$$

- $\lambda$: Regularization parameter  
- $\theta_j$: Model coefficients  
- $\lambda = 0$ ⇒ Linear Regression  
- $\lambda \to \infty$ ⇒ Coefficients shrink toward zero



### 🔹 Ridge Geometry

- Penalty region: **Circle**  
- Coefficients are shrunk but never exactly zero



### 🔹 When to Use Ridge?

- When all features are useful  
- When multicollinearity exists  
- When model overfits


## 🟧 Lasso Regression (L1 Regularization)

**Definition:** Lasso Regression adds an L1 penalty (absolute values of coefficients) to encourage sparsity in the model — i.e., it can set some coefficients to exactly 0.

### 🔹 Lasso Intuition:

**Cost Function:**

$$
J(\theta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{p} |\theta_j|
$$

- Encourages feature selection  
- Some coefficients may be set to zero  
- Ideal for sparse models



### 🔹 Lasso Geometry

- Penalty region: **Diamond**  
- Corners of diamond ⇒ sparsity (weights = 0)



### 🔹 When to Use Lasso?

- High-dimensional data  
- Feature selection needed  
- Few features are truly important  
- Risk of overfitting


### 🔹 Python Code for Lasso

```python

from sklearn.linear_model import Lasso

lasso = Lasso(alpha=0.1)
lasso.fit(X_train, y_train)

print("Lasso Coefficients:", lasso.coef_)


```


## 🟧 Ridge vs Lasso – Side-by-Side Comparison

| Property            | Ridge (L2)                 | Lasso (L1)                   |
|---------------------|----------------------------|------------------------------|
| Penalty term        | $\sum \theta^2$            | $\sum |\theta|$              |
| Shrinks coefficients| Yes                        | Yes                          |
| Sets coefficients to 0 | ❌ No                    | ✅ Yes                        |
| Feature selection   | ❌ No                       | ✅ Yes                        |
| Use case            | All features important      | Only few features important |
| Multicollinearity   | Handles well               | May select only one          |



## 🟧 ElasticNet Regression

**Definition:** ElasticNet is a hybrid of Ridge and Lasso that uses both L1 and L2 penalties to handle multicollinearity and do feature selection.

$$
J(\theta) = \text{SSE} + \lambda_1 \sum |\theta_j| + \lambda_2 \sum \theta_j^2
$$


### 🔹 Python Code for ElasticNet

```python

from sklearn.linear_model import ElasticNet

elastic = ElasticNet(alpha=1.0, l1_ratio=0.5)  # l1_ratio = mix between Lasso and Ridge
elastic.fit(X_train, y_train)


```

## 🟧 Interview Questions and Answers

#### **❓ Q1: What is the main difference between Ridge and Lasso Regression?**
**✅ Answer:**

 Ridge uses L2 penalty (squares of coefficients), shrinks coefficients but keeps all. Lasso uses L1 penalty (absolute values) and can shrink some to exactly zero, effectively performing feature selection.



#### **❓ Q2: Can Ridge regression set any coefficient to 0?**
**✅ Answer:**

 No. Ridge only reduces magnitudes of coefficients but doesn't set them exactly to zero.



#### **❓ Q3: What is ElasticNet and when is it used?**
**✅ Answer:**

 ElasticNet combines L1 and L2 regularization. It's used when:

- There are many correlated features  
- You need feature selection + multicollinearity handling



#### **❓ Q4: How to choose between Ridge, Lasso, and ElasticNet?**
**✅ Answer:**

| Use Case                          | Recommended   |
|----------------------------------|---------------|
| All features useful              | Ridge         |
| Few features useful (sparse)     | Lasso         |
| Correlated features + selection  | ElasticNet    |



#### **❓ Q5: How to tune $\lambda$ (alpha) in Ridge or Lasso?**
**✅ Answer:**

 Use `GridSearchCV` or `RandomizedSearchCV` from `scikit-learn` to find the best alpha ($\lambda$).

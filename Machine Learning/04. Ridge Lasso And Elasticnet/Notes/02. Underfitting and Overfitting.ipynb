# 📘 What is Underfitting and Overfitting?

---

## 🟧 Underfitting
**Definition**: A model is said to underfit when it is too simple to capture the underlying pattern of the data.

- Low training accuracy  
- Low test accuracy  
- Model is not learning enough from data

## 🟧 Overfitting
**Definition**: A model is said to overfit when it learns not only the patterns but also the noise in the training data.

- High training accuracy  
- Low test accuracy  
- Model memorizes instead of generalizing

## 🟧 Intuition with Real-life Example

### 🔹 Example: Predicting House Price
- **Underfitting**: You use only square footage to predict price, ignoring location, age, etc.  
- **Overfitting**: You use every single detail (like color of door or number of shelves) to predict price — your model memorizes the dataset.



## 🟧 Code Example: Underfitting vs Overfitting in Polynomial Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Generate synthetic data
import numpy as np
np.random.seed(0)
X = np.sort(np.random.rand(30, 1) * 2 - 1, axis=0)
y = X**3 + 0.3 * np.random.randn(30, 1)

# Try 3 models: Degree 1 (Underfit), Degree 3 (Good), Degree 10 (Overfit)
degrees = [1, 3, 10]
colors = ['red', 'green', 'blue']

plt.scatter(X, y, color='black', label='Data')

for d, c in zip(degrees, colors):
    poly = PolynomialFeatures(degree=d)
    X_poly = poly.fit_transform(X)
    model = LinearRegression().fit(X_poly, y)
    y_pred = model.predict(X_poly)
    plt.plot(X, y_pred, color=c, label=f'Degree {d}')
    
plt.legend()
plt.title("Underfitting vs Overfitting")
plt.show()
```



## 🟧 Bias-Variance Tradeoff

### 🔹 Definition:
- **Bias** is the error from overly simplistic assumptions  
- **Variance** is the error from sensitivity to small fluctuations

| Condition      | Bias | Variance | Generalization |
|----------------|------|----------|----------------|
| Underfitting   | High | Low      | Poor           |
| Good Fit       | Low  | Low      | Best           |
| Overfitting    | Low  | High     | Poor           |



## 🟧 How to Detect Underfitting or Overfitting?

| Check             | Underfitting | Overfitting |
|------------------|--------------|-------------|
| Training Error    | High         | Low         |
| Validation/Test Error | High     | High        |
| Gap (Train vs Test) | Small      | Large       |



## 🟧 How to Prevent Underfitting
- Use more complex models (e.g., polynomial regression)  
- Reduce regularization strength  
- Add more relevant features  
- Train longer (increase epochs)



## 🟧 How to Prevent Overfitting
- Use regularization (Ridge, Lasso)  
- Reduce model complexity (e.g., pruning in decision trees)  
- Cross-validation  
- Use more training data  
- Use early stopping (in neural networks)  
- Dropout (in deep learning)



## 🟧 Interview Questions and Answers

#### **❓ Q1: What is overfitting? How do you detect it?**
**✅ Answer:**

 Overfitting occurs when a model performs well on training data but poorly on test data. You can detect it by observing a high gap between training and validation accuracy.

#### **❓ Q2: What are some techniques to avoid overfitting?**
**✅ Answer:**

 Techniques include:  
- Regularization (L1, L2)  
- Cross-validation  
- Early stopping  
- Reducing model complexity  
- Data augmentation (for images)

#### **❓ Q3: Can increasing data help reduce overfitting?**
**✅ Answer:**

 Yes. More training data allows the model to generalize better and reduce its reliance on memorizing training examples.

#### **❓ Q4: What is the trade-off between bias and variance?**
**✅ Answer:**

 You must balance bias (underfitting) and variance (overfitting) to find a model that generalizes well. This is known as the bias-variance tradeoff.

#### **❓ Q5: Give real-world examples of underfitting and overfitting.**
**✅ Answer:**

- **Underfitting**: Using a constant or linear model to predict customer behavior in a complex market.  
- **Overfitting**: A neural network trained on small image datasets performs perfectly in training but fails on unseen data.

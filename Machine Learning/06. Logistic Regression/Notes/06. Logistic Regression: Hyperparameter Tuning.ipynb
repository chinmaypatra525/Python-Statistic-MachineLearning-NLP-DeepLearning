# 📘 Logistic Regression: Hyperparameter Tuning In-Depth



## 🟧 Why Tune Hyperparameters?

While Logistic Regression is a relatively simple model, it still has hyperparameters that significantly affect performance, especially:

- Regularization strength (C)  
- Regularization type (penalty)  
- Solver used for optimization  
- Class weight handling (for imbalanced datasets)

These hyperparameters control model complexity, prevent overfitting, and help improve generalization.



## 🟧 Key Hyperparameters in `LogisticRegression`

| Hyperparameter | Type       | Description                                                                 |
|----------------|------------|-----------------------------------------------------------------------------|
| C              | float      | Inverse of regularization strength. Smaller C means stronger regularization |
| penalty        | str        | Type of regularization: 'l1', 'l2', 'elasticnet', or 'none'                 |
| solver         | str        | Algorithm to use: 'liblinear', 'saga', 'lbfgs', 'newton-cg'                 |
| class_weight   | str or dict| Adjusts weights for imbalanced datasets: 'balanced' or custom dictionary   |
| max_iter       | int        | Maximum iterations for convergence                                          |
| fit_intercept  | bool       | Whether to learn an intercept term                                          |



## 🟧 1. C (Inverse Regularization Strength)

- High C → Less regularization → Risk of overfitting  
- Low C → More regularization → Risk of underfitting

**Formula (Cost Function):**

$$
J(\theta) = -\frac{1}{m} \sum [y \log(h_\theta) + (1 - y) \log(1 - h_\theta)] + \frac{\lambda}{2m} \sum \theta^2
$$

Here, $\lambda = \frac{1}{C}$

- Larger C → smaller penalty → fits training data closely



## 🟧 2. penalty

- `'l1'`: Lasso (feature selection, makes some weights = 0)  
- `'l2'`: Ridge (shrinks all weights)  
- `'elasticnet'`: combination of L1 and L2 (requires `solver = 'saga'`)



## 🟧 3. solver

| Solver      | Supports Penalty | Suitable For             |
|-------------|------------------|---------------------------|
| liblinear   | 'l1', 'l2'       | Small datasets            |
| saga        | All              | Large and sparse data     |
| lbfgs       | 'l2' only        | Default choice            |
| newton-cg   | 'l2'             | Multiclass problems       |



## 🟧 4. class_weight

Use `'balanced'` to adjust weights inversely proportional to class frequency.

Very helpful in imbalanced binary classification (e.g. fraud detection, churn)



## 🟧 5. Other Useful Parameters

- `max_iter`: Increase if solver fails to converge  
- `fit_intercept`: Whether to learn bias term

## 🟧 Tips for Tuning

- Use log scale for C values: [0.001, 0.01, 0.1, 1, 10, 100]  
- Combine `GridSearchCV` with `StratifiedKFold` for imbalanced data  
- Always check if model converges (increase `max_iter` if needed)



## 🟧 Interview Questions with Expert Answers

#### **❓ Q1: What does the C parameter control?**
**✅ Answer:**

 C is the inverse of regularization strength. Smaller C means more regularization, penalizing large weights to avoid overfitting. Large C fits training data better but risks overfitting.



#### **❓ Q2: When would you use L1 vs L2 regularization?**
**✅ Answer:**

- L1 (Lasso): Feature selection, makes some weights zero  
- L2 (Ridge): Shrinks all weights smoothly  

Use L1 for sparse feature sets and L2 for general regularization



#### **❓ Q3: How do you handle imbalanced classes in logistic regression?**
**✅ Answer:**

- Use `class_weight='balanced'`  
- Apply resampling (SMOTE/undersampling)  
- Use AUC, F1 score, or precision-recall as evaluation metrics



#### **❓ Q4: Why is liblinear often used in logistic regression?**
**✅ Answer:**

 It supports L1 and L2 penalties, is good for small datasets, and converges faster. However, it's not suited for large-scale or multi-class data.



#### **❓ Q5: How does regularization help in logistic regression?**
**✅ Answer:**

 Regularization prevents overfitting by penalizing large coefficients, thus improving generalization on unseen data.



#### **❓ Q6: How do you tune logistic regression for high recall?**
**✅ Answer:**

 Use `GridSearchCV` with `scoring='recall'` and try `class_weight='balanced'`. Also, manually adjust the threshold to favor true positives.

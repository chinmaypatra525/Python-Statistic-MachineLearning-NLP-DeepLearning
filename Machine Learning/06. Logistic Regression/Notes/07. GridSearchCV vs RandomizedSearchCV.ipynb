# 📘 Hyperparameter Tuning: GridSearchCV vs RandomizedSearchCV


---


## 🟧 Why Hyperparameter Tuning?

Every ML model has hyperparameters—configurable values not learned from data, but set before training (like `C`, `max_depth`, `n_estimators`, `learning_rate`, etc.).

We tune them to:

✅ Improve accuracy/generalization  
🚫 Avoid overfitting  
🔍 Optimize performance on unseen data



## 🟧 1. What is GridSearchCV?

Grid Search performs exhaustive search over a manually specified set of values for each hyperparameter.

**If you give:**

```python

C = [0.1, 1, 10], penalty = ['l1', 'l2']

```

It will train models for 3 x 2 = 6 combinations.

### ✅ Advantages:
- Simple and exhaustive  
- Finds global best in the grid

### ❌ Disadvantages:
- Very slow for large hyperparameter spaces  
- Inefficient for continuous values  
- Needs manual tuning of parameter range



## 🟧 2. What is RandomizedSearchCV?

Randomized Search samples a fixed number of parameter settings from given distributions.

Instead of searching every combination, it randomly tries `n_iter` combinations.

**Example:**
```python

from scipy.stats import uniform

param_dist = {
    'C': uniform(0.01, 10),      # Random floats between 0.01 and 10
    'penalty': ['l1', 'l2']
}

```

### ✅ Advantages:
- Much faster than GridSearchCV  
- Can search continuous ranges  
- Scalable to large datasets and many parameters

### ❌ Disadvantages:
- Might miss the absolute best  
- Random, so results vary slightly on each run



## 🟧 Comparison Table

| Feature                 | GridSearchCV       | RandomizedSearchCV        |
|------------------------|--------------------|----------------------------|
| Search Strategy         | Exhaustive         | Random Sampling            |
| Time Complexity         | Very high          | Lower                      |
| Best for                | Small search spaces| Large or infinite spaces   |
| Risk of Overfitting     | Higher if search space is large | Lower if well-tuned |
| Continuous Ranges       | ❌ Needs manual discretization | ✅ Can sample from distributions |
| Chance of Missing Optimum | Low (within grid) | Medium (due to sampling)   |


## 🟧 Best Practices

| Situation                      | Recommendation                          |
|-------------------------------|------------------------------------------|
| Small parameter space         | Use GridSearchCV                         |
| Large space or expensive models | Use RandomizedSearchCV                 |
| Want reproducibility          | Set `random_state`                       |
| Use validation curves         | To find ranges before grid               |
| Evaluation matters            | Use `scoring='f1'`, `roc_auc`, etc.     |
| Imbalanced data               | Use `StratifiedKFold` in `cv`           |



## 🟧 Interview Questions & Answers

#### **❓ Q1: Why use RandomizedSearch over GridSearch?**
**✅ Answer:**

 GridSearch is exhaustive and slow for large spaces. RandomizedSearch is more efficient and can explore continuous hyperparameter spaces without manual discretization.



#### **❓ Q2: Does RandomizedSearch always find the best hyperparameter?**
**✅ Answer:**

 Not always. Since it samples randomly, it might miss the exact global optimum, but it's much faster and often finds a near-optimal solution.



#### **❓ Q3: How can you improve RandomizedSearchCV performance?**
**✅ Answer:**

- Increase `n_iter`  
- Use informed distributions (like `loguniform`)  
- Combine with Bayesian optimization later (e.g. Optuna, Hyperopt)



#### **❓ Q4: When is GridSearch preferred?**
**✅ Answer:**

 When:

- You have few parameters  
- You want to ensure reproducibility  
- You want exhaustive tuning



#### **❓ Q5: What scoring metrics are used?**
**✅ Answer:**
  
Use `scoring=` argument:

- `'accuracy'` for balanced classes  
- `'f1'`, `'recall'`, `'precision'` for imbalanced  
- `'roc_auc'` for binary classification



#### **❓ Q6: How do you handle overfitting during hyperparameter tuning?**
**✅ Answer:**

- Use cross-validation (via `cv=`)  
- Choose proper scoring metric  
- Limit model complexity using regularization



## 🟧 Summary

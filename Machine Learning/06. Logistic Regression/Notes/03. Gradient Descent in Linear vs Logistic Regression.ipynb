# 📘 Gradient Descent in Linear vs Logistic Regression

---

## 🟧 What is Gradient Descent?

Gradient Descent is an optimization algorithm used to minimize the cost function by iteratively adjusting the model's weights ($w$) and bias ($b$) in the direction of steepest descent of the cost.



## 🟧 Update Rule:

- $w := w - \alpha \cdot \frac{\partial J}{\partial w}$
- $b := b - \alpha \cdot \frac{\partial J}{\partial b}$

Where:

- $J(w,b)$ is the cost function  
- $\alpha$ is the learning rate  



## 🟧 Cost Functions Compared

| Model              | Cost Function                | Convex? | Global Minima? |
|--------------------|------------------------------|---------|----------------|
| Linear Regression  | MSE (Mean Squared Error)     | Yes ✅  | Yes ✅         |
| Logistic Regression| Log Loss / Cross Entropy     | Yes ✅  | Yes ✅         |

> Both are convex functions, which means only one global minimum exists, no local minima traps.

## 🟧 Linear Regression + Gradient Descent

### 🔹 Hypothesis

$ \hat{y} = w^T x + b $

### 🔹 Cost Function: MSE

$ J(w,b) = \frac{1}{2n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2 $

### 🔹 Gradients

$ \frac{\partial J}{\partial w} = \frac{1}{n} \sum (w^T x_i + b - y_i) x_i $

$ \frac{\partial J}{\partial b} = \frac{1}{n} \sum (w^T x_i + b - y_i) $



## 🟧 Logistic Regression + Gradient Descent

### 🔹 Hypothesis

$ \hat{y} = \sigma(w^T x + b) \quad \text{where } \sigma(z) = \frac{1}{1 + e^{-z}} $

### 🔹 Cost Function: Binary Cross Entropy

$ J(w,b) = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right] $

✅ Convex ⇒ Only one global minimum  
✅ Smooth ⇒ Gradient descent works well

### 🔹 Gradients

$ \frac{\partial J}{\partial w} = \frac{1}{n} \sum (\hat{y}_i - y_i) x_i $

$ \frac{\partial J}{\partial b} = \frac{1}{n} \sum (\hat{y}_i - y_i) $



## 🟧 Why Only One Global Minimum in Logistic Regression?

The log-loss function is strictly convex with respect to $w$ and $b$ when features are linearly independent.

This means:

- 🧠 No plateaus  
- 🧠 No local minima  
- ✅ One global minimum → Gradient Descent is guaranteed to converge if learning rate is well-tuned.

## 🟧 Interview Q&A on Gradient Descent



### **❓ Q1: Why does logistic regression always converge to a global minimum?**
**✅ Answer:**

Because the log-loss function is convex, and convex functions have only one global minimum. So, if the learning rate is appropriate, gradient descent will converge to that minimum.



### **❓ Q2: What happens if the learning rate is too high?**
**✅ Answer:**
 
The updates may overshoot the minimum, causing the loss to diverge or oscillate. You won’t converge to the global minimum.



### **❓ Q3: Why can’t we use MSE as a cost function in logistic regression?**
**✅ Answer:**
 
MSE makes the loss non-convex in logistic regression due to the sigmoid function, leading to convergence issues. Log loss preserves convexity, ensuring stable convergence.



### **❓ Q4: What optimization variants exist beyond vanilla gradient descent?**
**✅ Answer:**

| Method        | Use Case                        |
|---------------|----------------------------------|
| SGD           | Online learning, stochastic updates |
| Mini-batch GD | Faster convergence              |
| Adam          | Adaptive learning rates         |
| Momentum      | Avoids local minima             |
| RMSProp       | Dampens oscillations            |



## 🟧 Best Practices to Ensure Global Convergence

- Normalize features: Speeds up convergence
- Use appropriate learning rate: Try $ \alpha = 0.01 $, tune further
- Use L2 regularization: Prevents overfitting
- Check for convergence: Stop when $ \Delta J $ is very small
- Use adaptive optimizers (Adam) for faster tuning

## 🟧 Python Code

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def gradient_descent_linear(X, y, lr=0.01, epochs=1000):
    m, n = X.shape
    w = np.zeros(n)
    b = 0
    for _ in range(epochs):
        y_pred = X @ w + b
        dw = (1/m) * X.T @ (y_pred - y)
        db = (1/m) * np.sum(y_pred - y)
        w -= lr * dw
        b -= lr * db
    return w, b

def gradient_descent_logistic(X, y, lr=0.01, epochs=1000):
    m, n = X.shape
    w = np.zeros(n)
    b = 0
    for _ in range(epochs):
        z = X @ w + b
        y_pred = sigmoid(z)
        dw = (1/m) * X.T @ (y_pred - y)
        db = (1/m) * np.sum(y_pred - y)
        w -= lr * dw
        b -= lr * db
    return w, b
```

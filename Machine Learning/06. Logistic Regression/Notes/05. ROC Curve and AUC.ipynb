## 🟧 Why ROC and AUC?

When dealing with classification problems like in Logistic Regression, SVM, Random Forest, etc., especially for imbalanced datasets, metrics like accuracy can be misleading.

To evaluate how good a model is at distinguishing between classes, we use ROC Curve and AUC (Area Under Curve).



## 🟧 What is ROC Curve?

ROC stands for Receiver Operating Characteristic.

A ROC Curve is a graphical plot that shows the diagnostic ability of a binary classifier system as its discrimination threshold is varied.



## 🟧 ROC Axes

**X-axis: False Positive Rate (FPR):**  
$FPR = \frac{FP}{FP + TN}$

**Y-axis: True Positive Rate (TPR) = Recall:**  
$TPR = \frac{TP}{TP + FN}$



## 🟧 ROC Curve Example

Each point on the curve represents a threshold

- Diagonal line → random model  
- Top-left curve → better model

## 🟧 How is ROC Curve Constructed?

Let’s say your classifier gives probability scores for 10 samples:

| Sample | True Label | Predicted Probability |
|--------|------------|------------------------|
| A      | 1          | 0.95                   |
| B      | 0          | 0.85                   |
| C      | 1          | 0.80                   |
| D      | 0          | 0.60                   |
| E      | 1          | 0.55                   |
| F      | 0          | 0.40                   |
| G      | 1          | 0.35                   |
| H      | 0          | 0.30                   |
| I      | 1          | 0.20                   |
| J      | 0          | 0.10                   |

You vary the threshold from 1.0 to 0.0 and compute TPR & FPR at each step → plot ROC.



## 🟧 What is AUC?

AUC stands for Area Under the ROC Curve.

AUC quantifies the overall ability of the classifier to distinguish between positive and negative classes.

AUC ranges from 0 to 1.



## 🟧 AUC Interpretation

| AUC Score | Interpretation            |
|-----------|---------------------------|
| 1.0       | Perfect model             |
| 0.9 - 1.0 | Excellent                 |
| 0.8 - 0.9 | Good                      |
| 0.7 - 0.8 | Fair                      |
| 0.6 - 0.7 | Poor                      |
| 0.5       | No discrimination (random guessing) |
| < 0.5     | Worse than random         |



## 🟧 TPR vs FPR Explained Visually

Let's break down the definitions:

**TPR (Recall):** Out of all actual positives, how many did we catch?

$TPR = \frac{TP}{TP + FN}$

**FPR:** Out of all actual negatives, how many were wrongly classified as positive?

$FPR = \frac{FP}{FP + TN}$



## 🟧 Multiple ROC Curves

Let’s compare three classifiers:

- Classifier A → close to top-left corner → High AUC  
- Classifier C → diagonal → Random predictions



## 🟧 Threshold and Tradeoff

As you lower the threshold:

- More positives are predicted → TPR increases  
- But also more negatives misclassified → FPR increases  

So, ROC shows how TPR and FPR change as you vary the threshold.



## 🟧 Why ROC is Better than Accuracy (in some cases)

- Accuracy is biased in imbalanced datasets  
- ROC considers true/false positives and negatives at all thresholds  
- Good when you care about ranking probabilities, not just hard labels



## 🟧 ROC vs PR Curve (Advanced Insight)

| Metric | ROC Curve | PR Curve         |
|--------|-----------|------------------|
| X-axis | FPR       | Recall           |
| Y-axis | TPR       | Precision        |
| Best for | Balanced data | Imbalanced data |
| Random baseline | Diagonal | Horizontal line |



## 🟧 When to Use ROC vs PR Curve?

- Use ROC when both classes are balanced  
- Use Precision-Recall (PR) when positive class is rare



## 🟧 Interview Questions and Answers

#### **❓ Q1. What is ROC Curve?**
**✅ Answer:**

 A ROC curve is a plot between True Positive Rate (Recall) and False Positive Rate, used to evaluate binary classifiers across different thresholds.



#### **❓ Q2. What does AUC signify?**
**✅ Answer:**

 AUC is the area under the ROC curve, indicating how well the classifier can separate positive and negative examples. AUC = 1 means perfect classification.



#### **❓ Q3. Why not just use accuracy?**
**✅ Answer:**

 Accuracy can be misleading, especially with imbalanced datasets. ROC and AUC provide a threshold-independent evaluation.



#### **❓ Q4. What’s the difference between ROC and Precision-Recall curve?**
**✅ Answer:**

 ROC shows TPR vs FPR and is useful for balanced datasets. PR Curve shows Precision vs Recall and is better for imbalanced datasets.



#### **❓ Q5. What if AUC < 0.5?**
**✅ Answer:**

 Model is performing worse than random. It might be reversing the predictions. Flip the labels and it may become useful.


```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Ground truth and predicted probabilities
y_true = [0, 0, 1, 1]
y_scores = [0.1, 0.4, 0.35, 0.8]

# Compute ROC curve and AUC
fpr, tpr, thresholds = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)

# Plotting
plt.figure()
plt.plot(fpr, tpr, color='blue', lw=2, label='ROC curve (AUC = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')  # random model
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('Receiver Operating Characteristic')
plt.legend()
plt.show()
```

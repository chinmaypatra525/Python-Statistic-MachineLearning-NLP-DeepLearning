## 🟧 1. What is Linear SVM?

### 🔹 Defination:

A **Linear SVM** tries to find a straight line (or hyperplane) that separates classes with maximum margin in the original input space.


### 🔹 Mathematically:

Find $w$ and $b$ such that:

$$
f(x) = w^T x + b
$$

And for classification:

$$
y_i(w^T x_i + b) \geq 1
$$


### 🔹 When to Use Linear SVM ?

- When data is linearly separable  
- When you have high-dimensional but sparse data  

**Examples:**
- Document classification  
- Spam filtering  
- Fraud detection (tabular)


## 🟧 2. What is Non-linear SVM?

### 🔹 Defination:

**Non-linear SVM** uses a kernel function to project data into a higher-dimensional space, where it becomes linearly separable.

### 🔹 Mathematically:

Instead of computing $w^T x + b$, it uses:

$$
f(x) = \sum_i \alpha_i y_i K(x_i, x) + b
$$

Where $K(x_i, x_j)$ is a kernel function like RBF, Polynomial, etc.


### 🔹 Common Kernels:

| Kernel    | Formula                                   | Use Case                       |
|-----------|--------------------------------------------|--------------------------------|
| RBF       | $\exp(-\gamma \|x - x'\|^2)$              | Complex, non-linear boundaries |
| Polynomial| $(x^T x' + c)^d$                          | Structured interactions        |
| Sigmoid   | $\tanh(\alpha x^T x' + c)$                | Rare, NN-like behavior         |



### 🔹 When to Use Non-linear SVM?

- When data shows clusters, spirals, curves  
- When decision boundary is not linear  

**Examples:**
- Image recognition  
- Speech classification  
- Medical diagnosis


### 🔹 Python Code

```python

clf = SVC(kernel='rbf', C=1, gamma='scale')
clf.fit(X_train, y_train)

```

## 🟧 Interview Questions & Answers

#### **❓1. What is the difference between Linear and Non-linear SVM?**
**✅ Answer:**

Linear SVM finds a straight line/hyperplane to separate classes.  
Non-linear SVM uses a kernel trick to project data into higher-dimensional space and find a linear separator there.



#### **❓2. When would you use a linear kernel instead of an RBF?**
**✅ Answer:**

- When the data is already linearly separable  
- Or when data is high-dimensional and sparse (e.g., text, TF-IDF)



#### **❓3. Why is non-linear SVM slower?**
**✅ Answer:**

Because the kernel matrix has complexity $O(n^2)$, which grows rapidly with dataset size.  
Also, more parameters (e.g., gamma, degree) require tuning.



#### **❓4. Can SVM automatically detect if data is linear or not?**
**✅ Answer:**

No. The choice of kernel is a design decision.  
Use cross-validation or domain knowledge to choose the best kernel.



#### **❓5. How to visualize the effect of linear vs non-linear kernel?**
**✅ Answer:**

Try SVM on a 2D toy dataset like `make_moons()` or `make_circles()` and plot decision boundaries for linear vs RBF kernels.
